```python
# Imports
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
```

```python
df = pd.read_csv("creditcard_sample.csv")
df.head()
```

```python
print("Shape:", df.shape)
print(df['Class'].value_counts())
print("Missing values:", df.isnull().sum().sum())
```

```python
scaler = StandardScaler()
df["Amount_scaled"] = scaler.fit_transform(df[["Amount"]])
df["Time_scaled"] = scaler.fit_transform(df[["Time"]])
```

```python
df = df.drop(columns=["Amount", "Time"])
```

```python
X = df.drop(columns=["Class"])
y = df["Class"]
```

```python
# Train-test split — stratify=y is IMPORTANT
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
```

```python
print("Train shape:", X_train.shape, "| Fraud in train:", y_train.sum())
print("Test shape:", X_test.shape, "| Fraud in test:", y_test.sum())
```

```python
# Step 3: SMOTE — sirf TRAINING data pe apply karo

smote = SMOTE(random_state=42)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", y_train.value_counts().to_dict())
print("After SMOTE:", y_train_sm.value_counts().to_dict())
print("New train shape:", X_train_sm.shape)
```

```python
# Step 4: Train 3 baseline models

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score, confusion_matrix

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=8, random_state=42, n_jobs=-1),
    "XGBoost": XGBClassifier(
        n_estimators=300, max_depth=5, learning_rate=0.1,
        eval_metric="logloss", random_state=42, n_jobs=-1
    ),
}

trained_models = {}
for name, model in models.items():
    model.fit(X_train_sm, y_train_sm)   # SMOTE-balanced data pe train
    trained_models[name] = model

    proba = model.predict_proba(X_test)[:, 1]   # fraud probability
    preds = model.predict(X_test)

    auc = roc_auc_score(y_test, proba)
    prec = precision_score(y_test, preds)
    rec = recall_score(y_test, preds)
    f1 = f1_score(y_test, preds)

    print(f"\n=== {name} ===")
    print(f"AUC: {auc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f}")
    print(confusion_matrix(y_test, preds))
```

```python
# Step 5A: Fuzzy Measure mu(S) — har coalition ke liye "capability score"

from itertools import combinations

model_names = ["Logistic Regression", "Random Forest", "XGBoost"]

# Chota validation split banao training data se (fuzzy measure estimate karne ke liye)
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train_sm, y_train_sm, test_size=0.2, stratify=y_train_sm, random_state=42
)

# Har model ko is chote X_fit pe fir se train karo (val set unbiased rahe isliye)
val_scores = {}
for name in model_names:
    m = trained_models[name]
    val_scores[name] = m.predict_proba(X_val)[:, 1]

def coalition_auc(subset):
    """Agar ye models ka simple average liya jaye, AUC kitna aata hai"""
    scores = np.mean([val_scores[m] for m in subset], axis=0)
    return roc_auc_score(y_val, scores)

mu = {}
mu[frozenset()] = 0.0
mu[frozenset(model_names)] = 1.0

# Singleton values (individual model ka weight)
singleton_aucs = {m: coalition_auc([m]) for m in model_names}
total = sum(singleton_aucs.values())
for m in model_names:
    mu[frozenset([m])] = 0.5 * (singleton_aucs[m] / total)

# Pair values (do models saath mein kitna better/synergy)
for pair in combinations(model_names, 2):
    pair_auc = coalition_auc(list(pair))
    best_individual = max(singleton_aucs[pair[0]], singleton_aucs[pair[1]])
    synergy = max(0.0, pair_auc - best_individual)
    base = mu[frozenset([pair[0]])] + mu[frozenset([pair[1]])]
    mu[frozenset(pair)] = min(0.95, base + synergy)

for k, v in sorted(mu.items(), key=lambda kv: len(kv[0])):
    label = "{}" if len(k)==0 else ", ".join(sorted(k))
    print(f"mu({{{label}}}) = {v:.4f}")
```

```python
# Step 5B: Choquet Integral aggregation

def choquet_integral(sample_scores: dict, mu: dict) -> float:
    """
    sample_scores = {model_name: probability} EK sample ke liye
    Ascending order mein sort karke, nested coalitions banate hain
    """
    sorted_models = sorted(sample_scores.items(), key=lambda kv: kv[1])
    names_sorted = [m for m, _ in sorted_models]
    vals_sorted = [v for _, v in sorted_models]

    result = 0.0
    prev_val = 0.0
    for i in range(len(vals_sorted)):
        coalition = frozenset(names_sorted[i:])   # ye model + isse upar wale sab
        weight = mu[coalition]
        result += (vals_sorted[i] - prev_val) * weight
        prev_val = vals_sorted[i]
    return result

def gsci_predict_proba(X, trained_models, mu, model_names):
    base_probs = {m: trained_models[m].predict_proba(X)[:, 1] for m in model_names}
    n = len(X)
    out = np.zeros(n)
    for i in range(n):
        sample_scores = {m: base_probs[m][i] for m in model_names}
        out[i] = choquet_integral(sample_scores, mu)
    return out

# Test set pe GSCI ensemble evaluate karo
gsci_test_proba = gsci_predict_proba(X_test, trained_models, mu, model_names)
gsci_preds = (gsci_test_proba >= 0.5).astype(int)

auc = roc_auc_score(y_test, gsci_test_proba)
prec = precision_score(y_test, gsci_preds)
rec = recall_score(y_test, gsci_preds)
f1 = f1_score(y_test, gsci_preds)

print(f"AUC: {auc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f}")
print(confusion_matrix(y_test, gsci_preds))
```

```python
# Step 6: Shapley Value of each base learner

import math

def shapley_value(model_name, model_names, mu):
    others = [m for m in model_names if m != model_name]
    n = len(model_names)
    total = 0.0
    for r in range(len(others) + 1):
        for subset in combinations(others, r):
            S = frozenset(subset)
            S_with = frozenset(subset) | {model_name}
            marginal = mu[S_with] - mu[S]   # ye model add karne se kitna faayda hua
            weight = (math.factorial(len(S)) * math.factorial(n - len(S) - 1)) / math.factorial(n)
            total += weight * marginal
    return total

print("=== Shapley Value (GSCI importance) ===")
for m in model_names:
    sv = shapley_value(m, model_names, mu)
    print(f"{m}: {sv:.4f}")
```

```python
# Step 7: SHAP explainability on XGBoost

import shap
import matplotlib.pyplot as plt

xgb_model = trained_models["XGBoost"]
explainer = shap.TreeExplainer(xgb_model)
shap_values = explainer.shap_values(X_test)

# Global feature importance
shap.summary_plot(shap_values, X_test, plot_type="bar", show=True)

# Beeswarm plot (feature values ka distribution + impact)
shap.summary_plot(shap_values, X_test, show=True)

# Top 10 features table
mean_abs_shap = np.abs(shap_values).mean(axis=0)
feature_importance = pd.DataFrame({
    "Feature": X_test.columns,
    "Mean |SHAP value|": mean_abs_shap
}).sort_values("Mean |SHAP value|", ascending=False).head(10)

print(feature_importance.to_string(index=False))
```

```python
"""
Trains the GSCI (fuzzy measure + Choquet integral) ensemble on the
creditcard fraud dataset and saves everything the Streamlit app needs
into a single joblib bundle: fraud_gsci_bundle.joblib

FIX vs original notebook: Amount and Time now get their OWN scaler
each (the original reused one StandardScaler object for both fits,
which silently overwrote the Amount scaling params with Time's).
"""

import math
from itertools import combinations

import joblib
import numpy as np
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, f1_score, precision_score, recall_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from xgboost import XGBClassifier

df = pd.read_csv("creditcard_sample.csv")  # or creditcard_sample.csv

# --- FIXED: separate scalers for Amount and Time ---
amount_scaler = StandardScaler()
time_scaler = StandardScaler()
df["Amount_scaled"] = amount_scaler.fit_transform(df[["Amount"]])
df["Time_scaled"] = time_scaler.fit_transform(df[["Time"]])
df = df.drop(columns=["Amount", "Time"])

X = df.drop(columns=["Class"])
y = df["Class"]
feature_order = list(X.columns)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)

smote = SMOTE(random_state=42)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)

model_names = ["Logistic Regression", "Random Forest", "XGBoost"]
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=8, random_state=42, n_jobs=-1),
    "XGBoost": XGBClassifier(n_estimators=300, max_depth=5, learning_rate=0.1,
                              eval_metric="logloss", random_state=42, n_jobs=-1),
}
trained_models = {}
for name, model in models.items():
    model.fit(X_train_sm, y_train_sm)
    trained_models[name] = model
    proba = model.predict_proba(X_test)[:, 1]
    preds = model.predict(X_test)
    print(f"{name}: AUC={roc_auc_score(y_test, proba):.4f} "
          f"P={precision_score(y_test, preds):.4f} R={recall_score(y_test, preds):.4f} "
          f"F1={f1_score(y_test, preds):.4f}")

# ---- Fuzzy measure mu(S) ----
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train_sm, y_train_sm, test_size=0.2, stratify=y_train_sm, random_state=42
)
val_scores = {name: trained_models[name].predict_proba(X_val)[:, 1] for name in model_names}

def coalition_auc(subset):
    scores = np.mean([val_scores[m] for m in subset], axis=0)
    return roc_auc_score(y_val, scores)

mu = {frozenset(): 0.0, frozenset(model_names): 1.0}
singleton_aucs = {m: coalition_auc([m]) for m in model_names}
total = sum(singleton_aucs.values())
for m in model_names:
    mu[frozenset([m])] = 0.5 * (singleton_aucs[m] / total)
for pair in combinations(model_names, 2):
    pair_auc = coalition_auc(list(pair))
    best_individual = max(singleton_aucs[pair[0]], singleton_aucs[pair[1]])
    synergy = max(0.0, pair_auc - best_individual)
    base = mu[frozenset([pair[0]])] + mu[frozenset([pair[1]])]
    mu[frozenset(pair)] = min(0.95, base + synergy)

# ---- Save everything the app needs ----
bundle = {
    "trained_models": trained_models,
    "mu": mu,
    "model_names": model_names,
    "feature_order": feature_order,
    "amount_scaler": amount_scaler,
    "time_scaler": time_scaler,
}
joblib.dump(bundle, "fraud_gsci_bundle.joblib")
print("\nSaved fraud_gsci_bundle.joblib")
```
